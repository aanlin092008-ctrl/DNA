```python
# DNA SEQUENCE CLASSIFICATION USING AI/ML
# Group A vs Group B

# STEP 1: Import libraries
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# STEP 2: Create the labelled DNA dataset
data = {
    "Sequence ID": ["S1", "S2", "S3", "S4"],
    "DNA Sequence": [
        "ATGCGATC",
        "ATGCGTAC",
        "GCGCGTAT",
        "GCGCGCGC"
    ],
    "Group": ["A", "A", "B", "B"]
}

df = pd.DataFrame(data)

# STEP 3: Extract numerical features
def extract_features(sequence):
    sequence = sequence.upper()

    A = sequence.count("A")
    T = sequence.count("T")
    G = sequence.count("G")
    C = sequence.count("C")

    length = len(sequence)

    if length == 0:
        gc_content = 0
    else:
        gc_content = ((G + C) / length) * 100

    return [A, T, G, C, length, gc_content]


# Apply feature extraction
features = df["DNA Sequence"].apply(extract_features)

feature_df = pd.DataFrame(
    features.tolist(),
    columns=["A", "T", "G", "C", "Length", "GC Content (%)"]
)

# Display feature table
print("DNA FEATURE TABLE")
print(pd.concat([df, feature_df], axis=1))


# STEP 4: Prepare training data
X = feature_df
y = df["Group"]


# STEP 5: Train the Decision Tree model
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

print("\nMODEL TRAINED SUCCESSFULLY!")


# STEP 6: Calculate model accuracy
predictions = model.predict(X)

accuracy = accuracy_score(y, predictions)

print("\nMODEL ACCURACY:", accuracy * 100, "%")


# STEP 7: Predict the unknown DNA sequence
unknown_sequence = "ATGCGCGATC"

unknown_features = extract_features(unknown_sequence)

unknown_df = pd.DataFrame(
    [unknown_features],
    columns=["A", "T", "G", "C", "Length", "GC Content (%)"]
)

predicted_group = model.predict(unknown_df)

print("\nUNKNOWN DNA SEQUENCE:", unknown_sequence)
print("EXTRACTED FEATURES:", unknown_features)
print("PREDICTED GROUP:", predicted_group[0])


# STEP 8: Display final result
print("\n----- FINAL RESULT -----")
print("Algorithm: Decision Tree Classifier")
print("Accuracy:", accuracy * 100, "%")
print("Unknown Sequence:", unknown_sequence)
print("Predicted Group:", predicted_group[0])
```
